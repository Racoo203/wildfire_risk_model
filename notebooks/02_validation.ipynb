```python
import pandas as pd
```

```python
# spring = pd.read_csv("./data/silver/model_datasets/dataset_train_spring.csv")
summer_train = pd.read_csv("./data/silver/model_datasets/dataset_train_summer.csv")
```

```python
def show_summary(df):
    temp_df = df.copy()
    # temp_df = temp_df.drop(columns=["label"], inplace = False).copy()
    summary_df = temp_df.describe().T
    summary_df["nan (%)"] = 100 * temp_df.isna().sum() / len(temp_df)
    return summary_df
```

```python
res = show_summary(summer_train)
```

```python
res[["min", "25%", "50%", "75%", "max"]].round(2)
```
